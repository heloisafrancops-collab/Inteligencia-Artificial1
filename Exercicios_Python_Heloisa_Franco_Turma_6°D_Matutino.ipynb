Para fazer a normalização utilizei a fórmula (array - menor) / (maior - menor). Exercício 19 — Ângulo entre dois arrays

Pergunta:
Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles usando o produto escalar normalizado.

---

Explicação:

---
Nesse exercício eu criei dois arrays de tamanho 20, que podem ser tratados como vetores. Primeiro utilizei np.dot() para calcular o produto escalar entre eles. Depois utilizei np.linalg.norm() para encontrar o tamanho de cada vetor. Com esses valores fiz o produto escalar normalizado, dividindo o produto escalar pelo produto dos tamanhos dos dois vetores. Esse resultado corresponde ao cosseno do ângulo entre eles. Depois utilizei np.arccos() para encontrar o ângulo e np.degrees() para converter o resultado de radianos para graus. No final mostrei os dois arrays e o ângulo encontrado.

```python
# Exercício 19 - Ângulo entre dois arrays

import numpy as np

# Criei dois arrays de tamanho 20
array1 = np.random.randint(1, 11, 20)
array2 = np.random.randint(1, 11, 20)

# Calculei o produto escalar
produto = np.dot(array1, array2)

# Calculei o tamanho de cada vetor
tamanho1 = np.linalg.norm(array1)
tamanho2 = np.linalg.norm(array2)

# Normalizei o produto escalar
cosseno = produto / (tamanho1 * tamanho2)

# Calculei o ângulo em radianos
angulo = np.arccos(cosseno)

# Converti o ângulo para graus
angulo_graus = np.degrees(angulo)

# Mostrei os arrays
print("Array 1:")
print(array1)

print("Array 2:")
print(array2)

# Mostrei o ângulo
print("Ângulo entre os arrays:", angulo_graus, "graus")
```

---
Nesse exercício eu utilizei o integrate do SciPy porque ele possui funções próprias para realizar cálculos de integração. Também utilizei o NumPy para conseguir trabalhar com o seno através do np.sin() e com o valor de π através do np.pi. Primeiro defini a função f(x) colocando exatamente a expressão que foi passada no exercício, onde x**2 representa x
2
 e np.sin(x) representa o seno de x. Depois utilizei o integrate.quad(), que é a função responsável por calcular a integral numericamente. Dentro dele passei a função que criei e os dois limites da integral: 0 como início e np.pi como final. O resultado do quad() retorna duas informações, então coloquei resultado para guardar o valor calculado da integral e erro para guardar a estimativa de erro do cálculo. Por último, utilizei o print() para mostrar essas duas informações na tela. dx
dy
	=x−y, com y(0)=1, no intervalo de 0 a 10

---
Nesse exercício eu utilizei o solve_ivp do SciPy porque ele é utilizado para resolver equações diferenciais numericamente. Primeiro defini a função equacao(x, y) colocando no return a expressão x - y, que corresponde à equação apresentada na questão. Depois utilizei o solve_ivp(), passando a função, o intervalo de 0 até 10 e [1] como condição inicial, já que o exercício informa que y(0)=1. O resultado da solução fica armazenado em solucao, e solucao.y[0][-1] pega o último valor encontrado para y, que corresponde ao final do intervalo. Exercício 8 — Linha com intervalo de confiança

Pergunta:
Plote um gráfico de linha com intervalos de confiança, usando sombreamento, para a função y = sin(x).

---

Explicação:

---
Nesse exercício eu criei a função seno e depois defini dois limites ao redor dela, um acima e outro abaixo. Utilizei fill_between() para preencher o espaço entre esses limites, criando o efeito de sombreamento pedido no exercício. A linha principal continua mostrando a função sin(x), enquanto a área sombreada representa o intervalo ao redor dela.

```python
# Exercício 8 - Intervalo de confiança

import matplotlib.pyplot as plt
import numpy as np

# Criei os valores de x
x = np.linspace(0, 10, 100)

# Calculei a função seno
y = np.sin(x)

# Criei os limites do intervalo
limite_superior = y + 0.2
limite_inferior = y - 0.2

# Criei a linha principal
plt.plot(x, y, label='Seno')

# Criei o sombreamento do intervalo
plt.fill_between(x, limite_inferior, limite_superior, alpha=0.3)

# Adicionei título e nomes aos eixos
plt.title("Função Seno com Intervalo")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")

# Mostrei o gráfico
plt.show()
```

Exercício 10 — Três linhas no mesmo gráfico

Pergunta:
Plote três gráficos de linha em um único gráfico, cada um com uma cor e estilo de linha diferentes.

---

Explicação:

---
Nesse exercício eu coloquei três séries diferentes no mesmo gráfico. Para cada uma utilizei plt.plot() e defini uma cor e um estilo de linha diferente. A primeira linha é contínua, a segunda é tracejada e a terceira é pontilhada. Dessa forma fica mais fácil diferenciar as três séries. O livro mostra que podemos fazer várias chamadas de plt.plot() para colocar várias séries no mesmo gráfico.

```python
# Exercício 10 - Três linhas

import matplotlib.pyplot as plt
import numpy as np

# Criei os valores de x
x = np.linspace(0, 10, 100)

# Criei três funções
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) + np.cos(x)

# Criei as três linhas
plt.plot(x, y1, color='red', linestyle='-')
plt.plot(x, y2, color='blue', linestyle='--')
plt.plot(x, y3, color='green', linestyle=':')

# Adicionei o título
plt.title("Três Gráficos de Linha")

# Mostrei o gráfico
plt.show()
```

Exercício 11 — Barras empilhadas

Pergunta:
Crie um gráfico de barras empilhadas para três categorias com três grupos de valores.

---

Explicação:

---
Nesse exercício eu utilizei três grupos de valores para montar as barras empilhadas. O primeiro grupo começa normalmente. Depois usei bottom=grupo1 para colocar o segundo grupo em cima do primeiro. Para o terceiro grupo, somei os valores dos dois primeiros e usei essa soma como bottom. Assim, os três grupos ficam acumulados na mesma barra. Essa é a mesma ideia utilizada no exemplo de barras empilhadas do material.

```python
# Exercício 11 - Barras empilhadas

import matplotlib.pyplot as plt

# Criei as categorias
categorias = ['A', 'B', 'C']

# Criei os três grupos
grupo1 = [3, 5, 4]
grupo2 = [2, 4, 3]
grupo3 = [1, 2, 2]

# Criei as barras
plt.bar(categorias, grupo1, label='Grupo 1')
plt.bar(categorias, grupo2, bottom=grupo1, label='Grupo 2')

# Somei os dois primeiros grupos para posicionar o terceiro
soma = [grupo1[i] + grupo2[i] for i in range(3)]

plt.bar(categorias, grupo3, bottom=soma, label='Grupo 3')

# Adicionei título e legenda
plt.title("Gráfico de Barras Empilhadas")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.legend()

# Mostrei o gráfico
plt.show()
```

Plote um gráfico de linha com uma função quadrática e ajuste os limites do eixo para focar nos valores x = -5 a x = 5.

Exercício 13 — Dispersão 3D

Pergunta:
Gere um gráfico de dispersão 3D para 100 pontos aleatórios utilizando Matplotlib.

---

Explicação:

---
Nesse exercício eu gerei 100 valores aleatórios para cada um dos três eixos. Como o gráfico possui três dimensões, utilizei projection='3d' para criar um eixo tridimensional. Depois utilizei scatter() para colocar os pontos nas posições de x, y e z. Por fim, identifiquei os três eixos para facilitar a leitura do gráfico.

```python
# Exercício 13 - Gráfico de dispersão 3D

import matplotlib.pyplot as plt
import numpy as np

# Criei 100 valores aleatórios
x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

# Criei a figura e o eixo 3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Criei o gráfico de dispersão
ax.scatter(x, y, z)

# Adicionei nomes aos eixos
ax.set_xlabel("Eixo X")
ax.set_ylabel("Eixo Y")
ax.set_zlabel("Eixo Z")

# Mostrei o gráfico
plt.show()
```

Crie um gráfico de barras horizontal para 4 categorias com valores gerados aleatoriamente.

Exercício 20 — Grade com espaçamento personalizado

Pergunta:
Adicione uma grade com espaçamento personalizado a um gráfico de linha para y = x².

---

Explicação:

---
Nesse exercício eu criei o gráfico da função x² e depois personalizei os pontos que aparecem nos eixos usando xticks() e yticks(). No eixo X coloquei uma distância de 2 entre os valores e no eixo Y uma distância de 10. Depois utilizei grid(True) para mostrar a grade seguindo esses valores. Assim consigo controlar melhor o espaçamento da grade e deixar o gráfico mais organizado.

```python
# Exercício 20 - Grade personalizada

import matplotlib.pyplot as plt
import numpy as np

# Criei os valores de x
x = np.linspace(-10, 10, 100)

# Criei a função x²
y = x ** 2

# Criei o gráfico
plt.plot(x, y)

# Adicionei título e nomes aos eixos
plt.title("Gráfico de y = x²")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")

# Defini os pontos da grade
plt.xticks(np.arange(-10, 11, 2))
plt.yticks(np.arange(0, 101, 10))

# Ativei a grade
plt.grid(True)

# Mostrei o gráfico
plt.show()
```
