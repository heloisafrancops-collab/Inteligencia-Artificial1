```python
# Exercício 8 - Verificação de Palíndromo

# Explicação:
# Nesse exercício eu utilizei o input() para receber uma palavra e comparei ela com a própria palavra invertida. Para inverter a palavra utilizei [::-1], onde os dois pontos (:) indicam que será percorrida toda a palavra e o -1 faz a leitura de trás para frente. Se a palavra original for igual à invertida, significa que ela é um palíndromo.

palavra = input("Digite uma palavra: ")  # Pede uma palavra ao usuário

if palavra == palavra[::-1]:  # Verifica se a palavra é igual quando escrita ao contrário
    print("É um palíndromo.")  # Mostra que a palavra é um palíndromo
else:  # Caso a condição não seja verdadeira
    print("Não é um palíndromo.")  # Mostra que não é um palíndromo
```

```python
# Exercício 9 - Fatorial de um Número

# Explicação:
# Nesse exercício eu utilizei o input() para receber um número digitado pelo usuário. Criei a variável fatorial com o valor 1, pois ela será responsável por armazenar o resultado das multiplicações. Depois utilizei o for com o range() para percorrer todos os números de 1 até o número informado. O "numero + 1" foi utilizado porque o range() não considera o último valor, então foi necessário adicionar 1 para que o número digitado também participasse do cálculo. A cada repetição, o valor de fatorial é multiplicado pelo valor de i. No final, utilizei o print() para mostrar o resultado.

numero = int(input("Digite um número: "))  # Pede um número e transforma em inteiro

fatorial = 1  # Começa o fatorial com 1

for i in range(1, numero + 1):  # Repete os números de 1 até o número informado
    fatorial = fatorial * i  # Multiplica o resultado pelo número atual

print(fatorial)  # Mostra o resultado do fatorial
```

```python
# Exercício 10 - Sequência de Fibonacci

# Explicação:
# Nesse exercício eu utilizei o input() para receber a quantidade de números
# da sequência que serão exibidos. Criei duas variáveis para armazenar os dois
# primeiros números da sequência (0 e 1). Depois utilizei o for para repetir
# o cálculo dos próximos números. A cada repetição, o próximo valor é encontrado
# somando os dois números anteriores, mostrando também a operação realizada.

quantidade = int(input("Digite a quantidade de números da sequência: "))  # Pede quantos números serão mostrados

numero1 = 0  # Define o primeiro número da sequência
numero2 = 1  # Define o segundo número da sequência

print(numero1)  # Mostra o primeiro número
print(numero2)  # Mostra o segundo número

for i in range(2, quantidade):  # Repete o cálculo a partir do terceiro número
    proximo = numero1 + numero2  # Soma os dois números anteriores para encontrar o próximo

    print(numero1, "+", numero2, "=", proximo)  # Mostra a soma que foi realizada

    numero1 = numero2  # Passa o segundo número para a primeira posição
    numero2 = proximo  # Passa o próximo número para a segunda posição
```

```python
# Exercício 12 - Contagem de Caracteres

# Explicação:
# Nesse exercício eu utilizei o input() para receber uma palavra ou frase. Depois utilizei o for para percorrer cada caractere digitado. O if verifica se o caractere já existe no dicionário. Se existir, ele adiciona mais 1 à quantidade. Caso contrário, ele cria esse caractere no dicionário com o valor 1. No final utilizei outro for para mostrar quantas vezes cada caractere apareceu.

texto = input("Digite uma palavra ou frase: ")  # Pede uma palavra ou frase ao usuário

contador = {}  # Cria um dicionário para guardar a quantidade de cada letra

for letra in texto:  # Percorre cada letra do texto
    if letra in contador:  # Verifica se a letra já está no dicionário
        contador[letra] = contador[letra] + 1  # Aumenta a quantidade da letra
    else:  # Caso a letra ainda não esteja no dicionário
        contador[letra] = 1  # Adiciona a letra começando com 1

for letra in contador:  # Percorre as letras que foram contadas
    print(letra, ":", contador[letra])  # Mostra cada letra e sua quantidade
```

```python
# Exercício 13 - Ordenação de Lista

# Explicação:
# Nesse exercício eu criei uma lista com alguns números e utilizei dois laços for para comparar os valores da lista. Quando um número era maior que o outro, troquei a posição deles. No final utilizei o print() para mostrar a lista em ordem crescente.

lista = [8, 3, 15, 1, 10, 6]  # Cria uma lista com números

for i in range(len(lista)):  # Percorre cada posição da lista
    for j in range(i + 1, len(lista)):  # Compara a posição atual com as próximas
        if lista[i] > lista[j]:  # Verifica se o número atual é maior que o próximo
            aux = lista[i]  # Guarda o número atual em uma variável auxiliar
            lista[i] = lista[j]  # Coloca o menor número na posição atual
            lista[j] = aux  # Coloca o número maior na outra posição

print(lista)  # Mostra a lista ordenada
```

```python
# Exercício 14 - Soma dos Elementos de uma Lista

# Explicação:
# Nesse exercício eu criei uma lista com alguns números e uma variável para armazenar a soma. Depois utilizei o for para percorrer todos os elementos da lista. A cada repetição, o valor do número é adicionado à variável soma. No final utilizei o print() para mostrar o resultado.

lista = [5, 10, 15, 20, 25]  # Cria uma lista com vários números

soma = 0  # Começa a soma com o valor 0

for numero in lista:  # Percorre cada número da lista
    soma = soma + numero  # Adiciona cada número ao valor da soma

print("A soma dos elementos da lista é:", soma)  # Mostra o resultado da soma
```

```python
# Exercício 15 - Média de uma Lista

# Explicação:
# Nesse exercício eu criei uma lista com alguns números e duas variáveis, uma para armazenar a soma dos elementos e outra para contar a quantidade de números da lista. Depois utilizei o for para percorrer todos os elementos, somando os valores e aumentando o contador. No final dividi a soma pela quantidade de elementos e utilizei o print() para mostrar a média.

lista = [5, 10, 15]  # Cria uma lista com os números

soma = 0  # Começa a soma com 0
quantidade = 0  # Começa a quantidade de números com 0

for numero in lista:  # Percorre cada número da lista
    soma = soma + numero  # Adiciona o número atual à soma
    quantidade = quantidade + 1  # Conta a quantidade de números da lista

media = soma / quantidade  # Calcula a média dividindo a soma pela quantidade

print("A média da lista é:", media)  # Mostra o resultado da média
```

```python
# Exercício 16 - Manipulação de Strings

# Explicação:
# Nesse exercício eu utilizei o input() para receber uma palavra ou frase. Depois utilizei [::-1] para inverter a string. Os dois pontos (:) indicam que será utilizada toda a string e o -1 faz a leitura de trás para frente. No final utilizei o print() para mostrar a string invertida.

texto = input("Digite uma palavra ou frase: ")  # Pede uma palavra ou frase ao usuário

invertida = texto[::-1]  # Inverte a palavra ou frase

print("String invertida:", invertida)  # Mostra o texto invertido
```

```python
# Exercício 17 - Dicionário de Contagem de Palavras

# Explicação:
# Nesse exercício eu utilizei o input() para receber uma frase. Depois utilizei o split() para separar a frase em palavras. Criei um dicionário para armazenar a quantidade de vezes que cada palavra aparece. Utilizei o for para percorrer todas as palavras e o if para verificar se a palavra já estava no dicionário. Se já existisse, aumentava a quantidade em 1. Caso contrário, ela era adicionada com o valor 1. No final utilizei outro for para mostrar o resultado.

frase = input("Digite uma frase: ")  # Pede uma frase ao usuário

palavras = frase.split()  # Separa a frase em palavras

contador = {}  # Cria um dicionário para contar as palavras

for palavra in palavras:  # Percorre cada palavra da frase
    if palavra in contador:  # Verifica se a palavra já foi contada
        contador[palavra] = contador[palavra] + 1  # Aumenta a quantidade da palavra
    else:  # Caso a palavra ainda não esteja no dicionário
        contador[palavra] = 1  # Adiciona a palavra começando com 1

for palavra in contador:  # Percorre as palavras que foram contadas
    print(palavra, ":", contador[palavra])  # Mostra a palavra e sua quantidade
```

```python
# Exercício 18 - Número Primo

# Explicação:
# Nesse exercício eu utilizei o input() para receber um número. Criei uma variável para contar quantas vezes esse número pode ser dividido sem deixar resto. Depois utilizei o for para testar todas as divisões de 1 até o número informado. Se a quantidade de divisões for igual a 2, significa que o número é primo. No final utilizei o print() para mostrar o resultado.

numero = int(input("Digite um número: "))  # Pede um número ao usuário

divisores = 0  # Começa a contagem de divisores com 0

for i in range(1, numero + 1):  # Percorre os números de 1 até o número informado
    if numero % i == 0:  # Verifica se o número é divisível pelo valor atual
        divisores = divisores + 1  # Conta mais um divisor

if divisores == 2:  # Verifica se o número possui exatamente dois divisores
    print("O número é primo.")  # Mostra que o número é primo
else:  # Caso tenha uma quantidade diferente de divisores
    print("O número não é primo.")  # Mostra que o número não é primo
```

```python
# Exercício 21 - Busca Binária

# Explicação:
# Nesse exercício eu utilizei uma lista ordenada e criei uma busca binária.
# A cada repetição, o programa verifica o elemento do meio da lista.
# Se o valor procurado for menor, a busca continua na metade esquerda.
# Se for maior, continua na metade direita.
# Assim a busca fica mais rápida, pois elimina vários valores a cada passo.

lista = [2, 5, 8, 12, 16, 20, 25, 30]  # Cria uma lista ordenada de números

numero = int(input("Digite o número que deseja encontrar: "))  # Pede o número que será procurado

inicio = 0  # Define o início da busca
fim = 7  # Define o fim da busca

encontrado = False  # Começa considerando que o número não foi encontrado

while inicio <= fim:  # Continua a busca enquanto houver posições para verificar
    meio = (inicio + fim) // 2  # Calcula a posição do meio da lista

    if lista[meio] == numero:  # Verifica se o número do meio é o número procurado
        encontrado = True  # Indica que o número foi encontrado
        break  # Encerra a busca

    elif numero < lista[meio]:  # Verifica se o número procurado é menor que o do meio
        fim = meio - 1  # Diminui o final da busca

    else:  # Caso o número seja maior que o valor do meio
        inicio = meio + 1  # Aumenta o início da busca

if encontrado:  # Verifica se o número foi encontrado
    print("Número encontrado na posição:", meio)  # Mostra a posição do número
else:  # Caso o número não tenha sido encontrado
    print("Número não encontrado")  # Mostra que o número não está na lista
```

```python
# Exercício 22 - Cifra de César

# Explicação:
# Nesse exercício eu utilizei a cifra de César para criptografar uma mensagem.
# Cada letra é substituída por outra que está algumas posições à frente no alfabeto.

mensagem = input("Digite uma mensagem: ")  # Pede uma mensagem ao usuário

deslocamento = int(input("Digite o deslocamento: "))  # Pede quantas posições as letras serão deslocadas

criptografada = ""  # Cria uma variável para guardar a mensagem criptografada

for letra in mensagem:  # Percorre cada letra da mensagem
    if letra.isalpha():  # Verifica se o caractere é uma letra
        codigo = ord(letra)  # Converte a letra para um código numérico
        nova_letra = chr(codigo + deslocamento)  # Cria uma nova letra com o deslocamento
        criptografada += nova_letra  # Adiciona a nova letra à mensagem
    else:  # Caso o caractere não seja uma letra
        criptografada += letra  # Mantém o caractere sem alteração

print("Mensagem criptografada:", criptografada)  # Mostra a mensagem criptografada
```

```python
# Exercício 23 - Descriptografar Cifra de César

# Explicação:
# Nesse exercício eu faço o processo contrário da criptografia.
# A letra volta a quantidade de posições informada pelo usuário.

mensagem = input("Digite a mensagem criptografada: ")  # Pede a mensagem que será descriptografada

deslocamento = int(input("Digite o deslocamento: "))  # Pede quantas posições as letras serão deslocadas

descriptografada = ""  # Cria uma variável para guardar a mensagem descriptografada

for letra in mensagem:  # Percorre cada letra da mensagem
    if letra.isalpha():  # Verifica se o caractere é uma letra
        codigo = ord(letra)  # Converte a letra para um código numérico
        nova_letra = chr(codigo - deslocamento)  # Volta a letra de acordo com o deslocamento
        descriptografada += nova_letra  # Adiciona a nova letra à mensagem
    else:  # Caso o caractere não seja uma letra
        descriptografada += letra  # Mantém o caractere sem alteração

print("Mensagem descriptografada:", descriptografada)  # Mostra a mensagem original
```

```python
# Exercício 24 - Validação de CPF

# Explicação:
# Nesse exercício eu utilizei o input() para receber um CPF informado pelo usuário.
# Primeiro, verifico se o CPF possui 11 números.
# Depois, realizo os cálculos dos dois dígitos verificadores usando multiplicações
# e pesos específicos. Por fim, comparo os dígitos calculados com os informados
# no CPF para verificar se ele é válido ou inválido.

cpf = input("Digite o CPF somente com números: ")  # Pede o CPF ao usuário

if len(cpf) == 11:  # Verifica se o CPF possui 11 números

    soma = 0  # Começa a soma com 0
    peso = 10  # Define o primeiro peso

    for numero in cpf[:9]:  # Percorre os 9 primeiros números do CPF
        soma += int(numero) * peso  # Multiplica cada número pelo seu peso e soma
        peso -= 1  # Diminui o peso em 1

    resto = soma % 11  # Calcula o resto da divisão da soma por 11

    if resto < 2:  # Verifica se o resto é menor que 2
        digito1 = 0  # Define o primeiro dígito como 0
    else:  # Caso o resto seja maior ou igual a 2
        digito1 = 11 - resto  # Calcula o primeiro dígito

    soma = 0  # Reinicia a soma
    peso = 11  # Define o peso inicial para o segundo dígito

    for numero in cpf[:9] + str(digito1):  # Percorre os 9 números e o primeiro dígito calculado
        soma += int(numero) * peso  # Multiplica cada número pelo seu peso e soma
        peso -= 1  # Diminui o peso em 1

    resto = soma % 11  # Calcula o resto da divisão da soma por 11

    if resto < 2:  # Verifica se o resto é menor que 2
        digito2 = 0  # Define o segundo dígito como 0
    else:  # Caso o resto seja maior ou igual a 2
        digito2 = 11 - resto  # Calcula o segundo dígito

    if int(cpf[9]) == digito1 and int(cpf[10]) == digito2:  # Compara os dígitos do CPF com os calculados
        print("CPF válido")  # Mostra que o CPF é válido
    else:  # Caso os dígitos sejam diferentes
        print("CPF inválido")  # Mostra que o CPF é inválido

else:  # Caso o CPF não tenha 11 números
    print("CPF inválido")  # Mostra que o CPF é inválido
```

```python
# Exercício 25 - Classificação de Números

# Explicação:
# Nesse exercício eu criei três listas vazias para armazenar os números
# positivos, negativos e iguais a zero.
#
# Depois, utilizei um laço de repetição for para percorrer todos os números
# da lista principal. A cada repetição, o programa verifica o valor do número
# usando estruturas condicionais.
#
# Se o número for maior que zero, ele é adicionado à lista de positivos.
# Se for menor que zero, ele é adicionado à lista de negativos.
# Caso o número seja igual a zero, ele é armazenado na lista de zeros.
#
# Ao final da execução, o programa exibe cada categoria separadamente,
# mostrando a classificação dos números informados.

numeros = [5, -3, 0, 8, -1, 0, 10]  # Cria uma lista com números positivos, negativos e zeros

positivos = []  # Cria uma lista para guardar os números positivos
negativos = []  # Cria uma lista para guardar os números negativos
zeros = []  # Cria uma lista para guardar os zeros

for numero in numeros:  # Percorre cada número da lista
    if numero > 0:  # Verifica se o número é positivo
        positivos += [numero]  # Adiciona o número na lista de positivos

    elif numero < 0:  # Verifica se o número é negativo
        negativos += [numero]  # Adiciona o número na lista de negativos

    else:  # Caso o número não seja positivo nem negativo
        zeros += [numero]  # Adiciona o zero na lista de zeros

print("Positivos:", positivos)  # Mostra os números positivos
print("Negativos:", negativos)  # Mostra os números negativos
print("Zeros:", zeros)  # Mostra os zeros
```

```python
# Exercício 30 - Jogo da Forca

# Explicação:
# Nesse exercício eu desenvolvi um jogo simples da forca. Primeiro,
# defini uma palavra secreta que deverá ser descoberta pelo usuário.
# Depois, criei uma variável para armazenar as letras que o jogador
# acertar e outra para controlar a quantidade de tentativas.
#
# Utilizei um laço de repetição para permitir que o usuário continue
# jogando enquanto ainda possuir tentativas e não descobrir a palavra.
# A cada rodada, o programa solicita uma letra e verifica se ela faz
# parte da palavra secreta.
#
# Se a letra estiver na palavra, ela é armazenada na variável que guarda
# os acertos. Caso contrário, o jogador perde uma tentativa.
#
# Em seguida, utilizei um laço for para percorrer cada letra da palavra.
# Se a letra já foi acertada, ela é exibida na tela. Caso contrário,
# é mostrado o caractere "_", indicando que aquela letra ainda não foi
# descoberta.
#
# O jogo termina quando o usuário descobrir todas as letras da palavra
# ou quando as tentativas acabarem. Ao final, o programa informa se o
# jogador venceu ou perdeu.

palavra = "python"  # Define a palavra que o jogador precisa descobrir

letras_acertadas = ""  # Guarda as letras que o jogador acertou

tentativas = 6  # Define a quantidade inicial de tentativas

while tentativas > 0:  # Continua o jogo enquanto ainda houver tentativas

    letra = input("Digite uma letra: ")  # Pede uma letra ao jogador

    if letra in palavra:  # Verifica se a letra está na palavra
        letras_acertadas += letra  # Adiciona a letra às letras acertadas
        print("Você acertou!")  # Mostra uma mensagem de acerto

    else:  # Caso a letra não esteja na palavra
        tentativas -= 1  # Diminui uma tentativa
        print("Você errou!")  # Mostra uma mensagem de erro
        print("Tentativas restantes:", tentativas)  # Mostra quantas tentativas ainda restam

    palavra_mostrada = ""  # Cria a palavra que será mostrada ao jogador

    for caractere in palavra:  # Percorre cada letra da palavra

        if caractere in letras_acertadas:  # Verifica se a letra já foi acertada
            palavra_mostrada += caractere  # Mostra a letra acertada

        else:  # Caso a letra ainda não tenha sido acertada
            palavra_mostrada += "_"  # Mostra um espaço no lugar da letra

    print("Palavra:", palavra_mostrada)  # Mostra a palavra com as letras descobertas

    if palavra_mostrada == palavra:  # Verifica se o jogador descobriu a palavra inteira
        print("Parabéns! Você venceu!")  # Mostra a mensagem de vitória
        break  # Encerra o jogo

if palavra_mostrada != palavra:  # Verifica se o jogador não conseguiu descobrir a palavra
    print("Você perdeu!")  # Mostra a mensagem de derrota
    print("A palavra era:", palavra)  # Mostra qual era a palavra
```

```python
# Exercício 31 - Ordenação por Bolha (Bubble Sort)

# Explicação:
# Nesse exercício eu utilizei o algoritmo Bubble Sort para ordenar uma
# lista de números em ordem crescente. O programa compara dois números
# vizinhos e, quando o número da esquerda é maior que o da direita,
# eles trocam de posição. A cada troca, a lista é exibida na tela para
# mostrar como a ordenação está acontecendo. No final, a lista fica
# totalmente organizada em ordem crescente.

numeros = [8, 3, 10, 1, 6]  # Cria a lista de números

print("Lista inicial:")  # Mostra o título da lista inicial
print(numeros)  # Mostra a lista antes da ordenação

for i in range(5):  # Repete o processo de comparação dos números
    for j in range(4):  # Percorre as posições da lista para fazer as comparações

        if numeros[j] > numeros[j + 1]:  # Verifica se o número atual é maior que o próximo
            aux = numeros[j]  # Guarda o número atual temporariamente
            numeros[j] = numeros[j + 1]  # Coloca o menor número na posição atual
            numeros[j + 1] = aux  # Coloca o maior número na próxima posição

            print(numeros)  # Mostra a lista após cada troca

print("Lista ordenada:")  # Mostra o título da lista ordenada
print(numeros)  # Mostra a lista depois da ordenação
```

```python
# Exercício 36 - Gerador de Senhas

# Explicação:
# Nesse exercício eu criei um gerador de senhas aleatórias utilizando letras
# e números. Primeiro, criei uma lista contendo todos os caracteres que podem
# ser usados na senha.
#
# Depois, utilizei o input() para receber o tamanho da senha que o usuário
# deseja criar. Em seguida, utilizei um laço de repetição for para escolher
# aleatoriamente cada caractere que fará parte da senha.
#
# A cada repetição, um novo caractere é escolhido e adicionado à variável
# que armazena a senha. Ao final, o programa exibe a senha gerada.

import random  # Importa a biblioteca para escolher caracteres aleatórios

caracteres = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"  # Define os caracteres que podem aparecer na senha

tamanho = int(input("Digite o tamanho da senha: "))  # Pede o tamanho da senha

senha = ""  # Cria uma variável vazia para guardar a senha

for i in range(tamanho):  # Repete de acordo com o tamanho escolhido

    caractere = random.choice(caracteres)  # Escolhe um caractere aleatório

    senha += caractere  # Adiciona o caractere escolhido à senha

print("Senha gerada:", senha)  # Mostra a senha gerada
```

```python
# Exercício 37 - Funções Recursivas

# Explicação:
# Nesse exercício eu criei uma função recursiva para calcular o fatorial
# de um número. Uma função recursiva é uma função que chama ela mesma
# durante sua execução.
#
# Primeiro, o programa recebe um número informado pelo usuário.
# Depois, a função verifica se o número é igual a 0 ou 1, pois nesses casos
# o resultado do fatorial é 1, encerrando a repetição.
#
# Caso o número seja maior que 1, a função multiplica o número pelo fatorial
# do número anterior, chamando ela mesma novamente até chegar ao caso final.
# Ao terminar os cálculos, o resultado do fatorial é exibido na tela.

def fatorial(numero):  # Cria uma função para calcular o fatorial

    if numero == 0 or numero == 1:  # Verifica se o número é 0 ou 1
        return 1  # Retorna 1, pois o fatorial de 0 e 1 é 1

    else:  # Caso o número seja maior que 1
        return numero * fatorial(numero - 1)  # Multiplica o número pelo fatorial do número anterior

numero = int(input("Digite um número: "))  # Pede um número ao usuário

resultado = fatorial(numero)  # Chama a função e guarda o resultado

print("Fatorial de", numero, "é:", resultado)  # Mostra o resultado do fatorial
```

```python
# Exercício 38 - Contar Palavras em Texto

# Explicação:
# Nesse exercício eu utilizei o input() para receber um texto digitado pelo
# usuário. Depois, utilizei a função split() para separar o texto em palavras.
# Cada palavra separada é armazenada dentro de uma lista.
#
# Em seguida, utilizei um laço de repetição for para passar por cada palavra
# da lista. A cada repetição, o contador aumenta 1, contando a quantidade de
# palavras existentes no texto. No final, o programa exibe o total de palavras.

texto = input("Digite um texto: ")  # Pede um texto ao usuário

palavras = texto.split()  # Separa o texto em palavras

contador = 0  # Começa a contagem de palavras com 0

for palavra in palavras:  # Percorre cada palavra do texto
    contador += 1  # Aumenta a contagem em 1

print("Quantidade de palavras:", contador)  # Mostra a quantidade de palavras
```

```python
# Exercício 40 - Lista de Compras

# Explicação:
# Nesse exercício eu criei um programa para gerenciar uma lista de compras.
# Primeiro, criei uma lista vazia onde os produtos adicionados pelo usuário
# serão armazenados.
#
# Depois, utilizei um laço de repetição para manter o programa funcionando,
# permitindo que o usuário escolha entre adicionar um item, remover um item
# ou visualizar a lista de compras.
#
# Quando o usuário escolhe adicionar, o produto é colocado na lista.
# Quando escolhe remover, o programa procura o item e remove da lista.
# Na opção listar, todos os produtos cadastrados são exibidos na tela.

compras = []  # Cria uma lista vazia para guardar os itens

opcao = 0  # Define a opção inicial

while opcao != 4:  # Continua o programa enquanto a opção não for 4

    print("\n1 - Adicionar item")  # Mostra a opção de adicionar
    print("2 - Remover item")  # Mostra a opção de remover
    print("3 - Listar compras")  # Mostra a opção de listar
    print("4 - Sair")  # Mostra a opção de sair

    opcao = int(input("Escolha uma opção: "))  # Pede ao usuário para escolher uma opção

    if opcao == 1:  # Verifica se o usuário escolheu adicionar

        item = input("Digite o item que deseja adicionar: ")  # Pede o item que será adicionado

        compras.append(item)  # Adiciona o item à lista

        print("Item adicionado!")  # Mostra uma mensagem de confirmação

    elif opcao == 2:  # Verifica se o usuário escolheu remover

        item = input("Digite o item que deseja remover: ")  # Pede o item que será removido

        if item in compras:  # Verifica se o item está na lista

            compras.remove(item)  # Remove o item da lista

            print("Item removido!")  # Mostra uma mensagem de confirmação

        else:  # Caso o item não esteja na lista

            print("Item não encontrado!")  # Mostra uma mensagem informando que o item não existe

    elif opcao == 3:  # Verifica se o usuário escolheu listar as compras

        print("Lista de compras:")  # Mostra o título da lista

        for item in compras:  # Percorre cada item da lista

            print(item)  # Mostra cada item da lista

    elif opcao == 4:  # Verifica se o usuário escolheu sair

        print("Programa encerrado!")  # Mostra a mensagem de encerramento

    else:  # Caso seja escolhida uma opção diferente das disponíveis

        print("Opção inválida!")  # Mostra uma mensagem de erro
```

```python
# Exercício 41 - Remover Duplicatas

# Explicação:
# Nesse exercício eu criei um programa que remove os números repetidos
# de uma lista. Primeiro, criei uma lista contendo alguns números,
# inclusive alguns repetidos. Depois, criei uma nova lista vazia para
# armazenar apenas os números que ainda não apareceram.
#
# Em seguida, utilizei um laço de repetição para percorrer todos os
# elementos da lista original. A cada repetição, o programa verifica
# se o número já está na nova lista. Se não estiver, ele é adicionado
# utilizando a concatenação de listas. Ao final, a nova lista contém
# apenas os números sem repetição.

# Lista original
numeros = [5, 2, 8, 2, 4, 5, 9, 8, 1]

# Lista sem duplicatas
sem_duplicatas = []

# Percorre a lista
for numero in numeros:

    # Verifica se o número ainda não existe na nova lista
    if numero not in sem_duplicatas:

        # Adiciona o número utilizando concatenação
        sem_duplicatas = sem_duplicatas + [numero]

print("Lista original:")
print(numeros)

print("Lista sem duplicatas:")
print(sem_duplicatas)
```

```python
# Exercício 42 - Números Primos até N

# Explicação:
# Nesse exercício eu utilizei o input() para receber um número informado
# pelo usuário. Depois, utilizei um laço de repetição para verificar todos
# os números de 2 até o número digitado. Para cada número, o programa verifica
# se ele possui algum divisor além de 1 e dele mesmo. Se não possuir,
# ele é considerado um número primo e é exibido na tela.

# Recebe do usuário o número limite
limite = int(input("Digite um número: "))

# Exibe uma mensagem informando o que será mostrado
print("Números primos até", limite, ":")

# Percorre todos os números de 2 até o número informado
for numero in range(2, limite + 1):

    # No início, considera que o número é primo
    primo = True

    # Testa todos os possíveis divisores
    for divisor in range(2, numero):

        # Verifica se a divisão é exata
        if numero % divisor == 0:

            # Se encontrar um divisor, o número não é primo
            primo = False

    # Se continuar sendo primo, exibe o número
    if primo == True:

        print(numero)
```

```python
# Exercício 44 - Fibonacci em Nível

# Explicação:
# Nesse exercício eu utilizei o input() para receber a quantidade de
# termos da sequência de Fibonacci que o usuário deseja visualizar.
# Depois, criei duas variáveis para armazenar os dois primeiros números
# da sequência. Em seguida, utilizei um laço de repetição para gerar
# a quantidade de termos informada pelo usuário. A cada repetição,
# o programa exibe o número atual, calcula o próximo número da sequência
# e atualiza os valores para continuar o processo.

# Recebe a quantidade de termos da sequência
quantidade = int(input("Digite a quantidade de termos: "))

# Primeiro número da sequência
anterior = 0

# Segundo número da sequência
atual = 1

# Exibe uma mensagem
print("Sequência de Fibonacci:")

# Repete a quantidade de vezes informada
for i in range(quantidade):

    # Exibe o número atual da sequência
    print(anterior)

    # Calcula o próximo número
    proximo = anterior + atual

    # Atualiza o valor do número anterior
    anterior = atual

    # Atualiza o valor do número atual
    atual = proximo
```

```python
# Exercício 45 - Par ou Ímpar em Lista

# Explicação:
# Nesse exercício eu utilizei uma lista de números e percorri todos os
# seus elementos utilizando um laço de repetição. Para cada número,
# utilizei o operador % (resto da divisão) para verificar se ele é
# par ou ímpar.
#
# Se o resto da divisão por 2 for igual a 0, o número é par.
# Caso contrário, o número é ímpar. Ao final, o programa informa a
# classificação de cada número da lista.

# Lista de números
numeros = [8, 3, 10, 7, 12, 5, 18, 9]

# Percorre todos os números da lista
for numero in numeros:

    # Verifica se o número é par
    if numero % 2 == 0:

        # Exibe que o número é par
        print(numero, "é par")

    # Caso contrário, o número é ímpar
    else:

        # Exibe que o número é ímpar
        print(numero, "é ímpar")
```

```python
# Exercício 46 - Calcular Média e Desvio Padrão

# Explicação:
# Nesse exercício eu utilizei uma lista de números para calcular a média
# e o desvio padrão. Primeiro, somei todos os números da lista e dividi
# pela quantidade de elementos para encontrar a média.
#
# Depois, calculei a diferença entre cada número e a média. Essa diferença
# foi elevada ao quadrado e somada. Em seguida, dividi essa soma pela
# quantidade de números para encontrar a variância.
#
# Por último, calculei a raiz quadrada da variância utilizando o operador
# de potência (** 0.5), encontrando o desvio padrão.

# Lista de números
numeros = [5, 8, 10, 6, 9]

# Variável para armazenar a soma dos números
soma = 0

# Variável para contar a quantidade de números
quantidade = 0

# Percorre toda a lista
for numero in numeros:

    # Soma os números
    soma = soma + numero

    # Conta quantos números existem
    quantidade = quantidade + 1

# Calcula a média
media = soma / quantidade

# Exibe a média
print("Média:", media)

# Variável para armazenar a soma das diferenças ao quadrado
soma_diferencas = 0

# Percorre novamente todos os números
for numero in numeros:

    # Calcula a diferença entre o número e a média
    diferenca = numero - media

    # Eleva a diferença ao quadrado
    quadrado = diferenca ** 2

    # Soma os quadrados
    soma_diferencas = soma_diferencas + quadrado

# Calcula a variância
variancia = soma_diferencas / quantidade

# Calcula o desvio padrão (raiz quadrada da variância)
desvio = variancia ** 0.5

# Exibe o desvio padrão
print("Desvio padrão:", desvio)
```

```python
# Exercício 47 - Contar Números em Lista

# Explicação:
# Nesse exercício eu utilizei uma lista de números e contei quantas vezes
# cada número aparece. Primeiro, percorri toda a lista. Para evitar contar
# o mesmo número mais de uma vez, verifiquei se ele já havia sido contado.
# Se ainda não tivesse sido contado, percorri novamente a lista para contar
# quantas vezes ele aparecia. No final, exibi o número e a quantidade de
# vezes que ele foi encontrado.

# Lista de números
numeros = [2, 5, 2, 8, 5, 2, 9, 8, 1]

# Lista para guardar os números que já foram contados
contados = []

# Percorre todos os números da lista
for numero in numeros:

    # Verifica se o número ainda não foi contado
    if numero not in contados:

        # Variável para contar quantas vezes o número aparece
        contador = 0

        # Percorre novamente toda a lista
        for valor in numeros:

            # Verifica se o número é igual ao valor da lista
            if numero == valor:

                # Aumenta o contador
                contador = contador + 1

        # Exibe o resultado
        print(numero, "aparece", contador, "vez(es).")

        # Guarda o número para não contá-lo novamente
        contados = contados + [numero]
```

```python
# Exercício 48 - Ordenação por Inserção (Insertion Sort)

# Explicação:
# Nesse exercício eu utilizei o algoritmo Insertion Sort para ordenar
# uma lista de números em ordem crescente.
#
# O algoritmo funciona como se estivéssemos organizando cartas na mão.
# Ele começa considerando que o primeiro número já está na posição correta.
# Depois, pega o próximo número e o compara com os anteriores.
# Se o número for menor, os outros números são deslocados para a direita
# até que seja encontrada a posição correta para inseri-lo.
#
# Esse processo é repetido até que toda a lista esteja ordenada.

# Lista de números
numeros = [8, 3, 10, 1, 6]

# Exibe a lista antes da ordenação
print("Lista inicial:")
print(numeros)

# Percorre a lista a partir do segundo elemento
for i in range(1, len(numeros)):

    # Guarda o número que será inserido
    atual = numeros[i]

    # Guarda a posição anterior
    j = i - 1

    # Enquanto existir posição anterior
    # e o número anterior for maior que o atual
    while j >= 0 and numeros[j] > atual:

        # Desloca o número maior uma posição para a direita
        numeros[j + 1] = numeros[j]

        # Vai para a posição anterior
        j = j - 1

    # Coloca o número na posição correta
    numeros[j + 1] = atual

    # Exibe a lista após cada inserção
    print(numeros)

# Exibe a lista ordenada
print("Lista ordenada:")
print(numeros)
```

```python
# Exercício 49 - Soma dos Dígitos

# Explicação:
# Nesse exercício eu utilizei o input() para receber um número inteiro
# informado pelo usuário. Depois, percorri cada dígito desse número
# utilizando um laço de repetição.
#
# A cada repetição, o programa converte o dígito para inteiro e soma
# esse valor em uma variável acumuladora. No final, o programa exibe
# a soma de todos os dígitos do número.

# Recebe um número do usuário
numero = input("Digite um número inteiro: ")

# Variável que armazenará a soma dos dígitos
soma = 0

# Percorre cada dígito do número
for digito in numero:

    # Converte o dígito para inteiro e soma
    soma = soma + int(digito)

# Exibe o número digitado
print("Número digitado:", numero)

# Exibe a soma dos dígitos
print("Soma dos dígitos:", soma)
```

```python
# Exercício 50 - Cálculo de Frequência de Letras

# Explicação:
# Nesse exercício eu utilizei o input() para receber uma palavra ou frase
# digitada pelo usuário. Depois, percorri cada letra do texto para descobrir
# quantas vezes ela aparece.
#
# Para não contar a mesma letra mais de uma vez, utilizei uma lista para
# armazenar as letras que já foram verificadas. Em seguida, percorri o texto
# novamente para contar quantas vezes cada letra aparece.
#
# No final, o programa exibe cada letra e a quantidade de vezes que ela foi
# encontrada no texto.

# Recebe uma palavra ou frase
texto = input("Digite uma palavra ou frase: ")

# Lista para guardar as letras que já foram contadas
letras_contadas = []

# Percorre cada caractere do texto
for letra in texto:

    # Ignora os espaços
    if letra != " ":

        # Verifica se a letra ainda não foi contada
        if letra not in letras_contadas:

            # Variável para contar quantas vezes a letra aparece
            contador = 0

            # Percorre novamente todo o texto
            for caractere in texto:

                # Verifica se a letra é igual ao caractere
                if letra == caractere:

                    # Aumenta o contador
                    contador = contador + 1

            # Exibe o resultado
            print(letra, "aparece", contador, "vez(es).")

            # Guarda a letra para não contá-la novamente
            letras_contadas = letras_contadas + [letra]
```
